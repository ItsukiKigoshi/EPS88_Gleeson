## When estimating earthquake probabilities it is common to consider ${\Delta}time=30 yrs$. Why is it?
Because it is engineering demand for building code purposes (Gerstenberger et al., 2020).

## Create Dictionary of the Estimated Numbers of the Earthquakes with more than Magnitudes `[4,5,6,7]` based on Gutenberg-Richter Relationship

````python
# dictionary comprehension is awesome! https://www.geeksforgeeks.org/python/python-dictionary-comprehension/
N_annual = {m: 10**(3.68 - 0.82 * m) for m in [4, 5, 6, 7]}
````

## Calculate the probability of at least 1 earthquake happens with Magnitudes `[4,5,6,7]` within 10, 30 yrs.
I will use the Poisson Distribution as stated in Week 5 to calculate the probability.

````python
dt = 30
expected_val_30 = {m: rate * dt for m, rate in N_annual.items()}
print(expected_val_30)
````

````python
# Function from Week 5
import numpy as np
import math
def poisson_probability(k,lam):
    """
    This function computes the probability of getting k particular outcomes when the expected rate is lam.
    """
    # compute the poisson probability of getting k outcomes when the expected rate is lam
    prob = (np.exp(-1*lam))*(lam**k)/math.factorial(k)

    #return the output
    return prob
````

````python
print("The probability of one or more earthquakes with the shown magnitude in the next 30 years:")
for m, mu in expected_val_30.items():
    prob_1_plus = 1 - poisson_probability(0, mu)
    print(f"M{m} or bigger: {prob_1_plus:.2%}")
````

````python
dt = 10
expected_val_10 = {m: rate * dt for m, rate in N_annual.items()}
print(expected_val_10)
````

````python
print("The probability of one or more earthquakes with the shown magnitude in the next 10 years:")
for m, mu in expected_val_10.items():
    prob_1_plus = 1 - poisson_probability(0, mu)
    print(f"M{m} or bigger: {prob_1_plus:.2%}")
````

## Calculate the probability of 3 or more earthquakes happens with Magnitudes `[4,5,6,7]` within 10, 30 yrs.
Here, I will suppose that the problem wants us to derive the probability of 3 or more (instead of "2 or more") earthquakes as the problem states ("> 2").

````python
def probability_more_than_2(mu):
    p0 = poisson_probability(0, mu)
    p1 = poisson_probability(1, mu)
    p2 = poisson_probability(2, mu)

    prob = 1 - (p0 + p1 + p2)
    return prob
````

````python
for m, mu in expected_val_30.items():
    p_multiple = probability_more_than_2(mu)
    print(f"the probability of multiple earthquakes (>2) happening within 30 yrs: {p_multiple:.2%}")
````

````python
for m, mu in expected_val_10.items():
    p_multiple = probability_more_than_2(mu)
    print(f"the probability of multiple earthquakes (>2) happening within 10 yrs: {p_multiple:.2%}")
````

# Result

````python
# Result
magnitudes = [4, 5, 6, 7]

# <, >, and numbers can specify spaces!: https://www.geeksforgeeks.org/python/string-alignment-in-python-f-string/
print(f"{'Mag':<4} | {'Period':<10} | {'Prob (1+)':>12} | {'Prob (>2)':>12}")

for m in magnitudes:
    print("-"*48) # 48 == 4 + 10 + 12 + 12

    # 30 years
    mu30 = expected_val_30[m]
    p1_30 = 1 - poisson_probability(0, mu30)
    pm_30 = probability_more_than_2(mu30)
    print(f"M{m:<3} | {'30 years':<10} | {p1_30:>12.2%} | {pm_30:>12.2%}")

    # 10 years
    mu10 = expected_val_10[m]
    p1_10 = 1 - poisson_probability(0, mu10)
    pm_10 = probability_more_than_2(mu10)
    print(f"{'':<4} | {'10 years':<10} | {p1_10:>12.2%} | {pm_10:>12.2%}")

````
