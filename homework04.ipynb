### Defining a function

When you may be doing a calculation more than once it is a good idea to define a function. Let's define a function that will take an east magnitude and a north magnitude and return a direction between 0 and 360.

```python
def GPS_direction(east_magnitude, north_magnitude):
    direction_rad = np.arctan2(east_magnitude, north_magnitude)
    direction = np.rad2deg(direction_rad) % 360
    return direction
```

Are the slopes of the two periods siginficantly different from each other? Is there a change in the land subsidence rate through time?

Remember (from lectures), the standard error of the slope for linear regression is:

$$ SE_{s} = \sqrt{\sigma^2 / \sum{(x_{i} - \bar{x})}^2} $$

where $\sigma^2 = \sum{(y_i - \hat{y}_i)^2}/(n-p) $ and $n$ is the number of observations, $p$ is the number of predictors. Together $n - p$ equals the 'degrees of freedom' $dof$.

The standard error of the intercept can then be given by:

$$ SE_{I} = \sqrt{\sigma^2(1/n + \bar{x}^2/\sum{(x_i - \bar{x})**2})} $$

Remember $\bar{x}$ indicates the mean value of $x$ and $\hat{y}_i$ indicates the predicted value of $y$ at position $i$.

```python
X = P566_GPS_18_23['days']
n = len(X)
p = len(model_18_23.coef_) + 1
dof = n - p
y = P566_GPS_18_23['Vertical (mm)']
y_pred = model_18_23.predict(X.values.reshape(-1,1))

sigma = np.sum((y - y_pred)**2) /dof

se_slope = np.sqrt(sigma/np.sum((X - X.mean())**2))
se_inter = np.sqrt(sigma * (1/n + (X.mean()**2)/np.sum((X - X.mean())**2)))

print(f"Standard error of slope: {se_slope:.4f}")
print(f"Standard error of intercept: {se_inter:.4f}")
```
